```python
from core import Builder, VectorStore, embed, load
from core.ui import chat_print, init_style, loading, show_history
from dotenv import load_dotenv
from IPython.display import HTML
```

```python
load_dotenv()
init_style()
```

```python
documents = load("./rules.txt")
embeddings = embed(texts=documents)
vector_store = VectorStore(documents=documents, embeddings=embeddings)
```

```python
VERBOSE = True
builder = Builder(retrieved_documents=[])

while True:
    user_query = input("").strip()

    if user_query == "q":
        break

    chat_print(text=user_query, role="user")
    loader = loading()

    query_embedding = embed(texts=[user_query])
    builder.retrieved_documents = vector_store.search(query_embedding[0])

    answer = builder.answer(query=user_query)
    if loader is not None:
        loader.update(HTML(""))

    if VERBOSE:
        # found_rules(retrieved_documents=retrieved_documents)

        show_history(
            messages=builder.get_messages(),
            title="لیست پیام‌های ارسالی به مدل",
        )

    chat_print(text=answer, role="assistant")
```
