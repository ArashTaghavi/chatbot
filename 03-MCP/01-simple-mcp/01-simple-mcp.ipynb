### With MCP SERVER

```python
client = MultiServerMCPClient(
    {
        "my_server": {
            "transport": "streamable_http",
            "url": "http://127.0.0.1:8000/mcp",
        }
    }
)

tools = await client.get_tools()

print([tool.name for tool in tools])
```

```python
from dotenv import load_dotenv
from langchain.chat_models import init_chat_model

load_dotenv()

model = init_chat_model(
    model="gpt-4o",
    model_provider="openai",
    temperature=0,
    timeout=30,
    max_retries=3,
)
```

```python
from langchain.agents import create_agent

# model همان مدل LangChain است که قبلاً تنظیم کرده‌ای

agent = create_agent(
    model=model,
    tools=tools,
    system_prompt="تو یک دستیار فارسی هستی. از ابزارهای MCP استفاده کن.",
)
```

```python
result = await agent.ainvoke({"messages": [{"role": "user", "content": "ساعت چنده"}]})

print(result["messages"][-1].content)
```

### Use Langchain

```python
from datetime import datetime

from langchain.agents import create_agent
from langchain.tools import tool


@tool
def say_hello(name: str) -> str:
    """Say hello to a person."""
    return f"Hello {name}"


@tool
def add_numbers(a: int, b: int) -> int:
    """Add two numbers."""
    return a + b


@tool
def get_current_time() -> str:
    """Get current server date and time."""
    return datetime.now().strftime("%Y-%m-%d %H:%M:%S")


agent = create_agent(
    model=model,
    tools=[
        say_hello,
        add_numbers,
        get_current_time,
    ],
)

result = agent.invoke({"messages": [{"role": "user", "content": "ساعت چنده"}]})
# result = await agent.ainvoke({"messages": [{"role": "user", "content": "ساعت چنده"}]})

result['messages'][-1].content
```
